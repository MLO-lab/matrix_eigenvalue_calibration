## Preprocessing TriviaQA dataset into an evaluation split

```python
import json, os, random
```

```python
with open('../data/raw/TriviaQA/unfiltered-web-dev.json', 'r') as file:
    data = json.load(file)
    
```

```python
for i, item in enumerate(data['Data']):
    item['datapoint_id']= i
```

```python
len(data['Data'])
```

```python
random.seed(42)
```

```python
sample = random.sample(data['Data'], 2000)
```

```python
sample[0]
```

```python
eval_data = []
for i, item in enumerate(sample):
    eval_data.append({
        'question_id': i,
        'datapoint_id': item['datapoint_id'],
        'question': item['Question'],
        'gt_answer': item['Answer']['Value'],
    })
```

```python
eval_data
```

```python
with open('../data/eval/TriviaQA_2k.json', 'w') as file:
    json.dump(eval_data, file) 
```

```python

gt_answers = []
for entry in eval_data:
    gt_answers.append({
        "question_id": entry["question_id"],
        "variant_id": 0,
        "answer_id": 0,
        "generated_answer": entry["gt_answer"]
    })
with open('../data/logs/TriviaQA_2k/gt-answers.json', 'w') as f:
    json.dump(gt_answers, f, indent=4)
```
